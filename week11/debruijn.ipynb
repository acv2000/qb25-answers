```python
reads = ['ATTCA', 'ATTGA', 'CATTG', 'CTTAT', 'GATTG', 'TATTT',
         'TCATT', 'TCTTA', 'TGATT', 'TTATT', 'TTCAT', 'TTCTT', 'TTGAT']
k = 3

# Edges
graph = set()
for read in reads:
    for i in range(len(read) - k):
        kmer1 = read[i:i+k]
        kmer2 = read[i+1:i+1+k]
        edge = f"{kmer1} -> {kmer2}"
        graph.add(edge)

# DOT file
with open("debruijn_graph.dot", "w") as f:
    f.write("digraph G {\n")
    for edge in sorted(graph):
        f.write(f"    {edge};\n")
    f.write("}\n")
```
