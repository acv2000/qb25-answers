```python
# Step 1.2
import random
import math
import numpy as np
from scipy.stats import poisson, norm
```

```python
# Chosen Parameters 
genomesize = 1_000_000     
readlength = 100            
coverage = 3                

# Number of reads needed
num_reads = int((coverage * genomesize) / readlength)
print("Number of reads:", num_reads)

# Coverage
genome_coverage = np.zeros(genomesize, dtype=int)

# Simulate possible reads 
for i in range(num_reads):
    startpos = random.randint(0, genomesize - readlength)
    genome_coverage[startpos : startpos + readlength] += 1

print("Simulation complete.")
print("Max observed coverage:", genome_coverage.max())

# Save coverage 
with open("coverage.txt", "w") as f:
    for c in genome_coverage:
        f.write(str(c) + "\n")

print("Coverage saved to coverage.txt")

# Distribution estimates 
xs = range(0, genome_coverage.max() + 1)

# Poisson PMF
poisson_estimates = [poisson.pmf(x, coverage) for x in xs]

# Normal PDF (approximation to Poisson for large n)
normal_estimates = [
    norm.pdf(x, loc=coverage, scale=math.sqrt(coverage)) for x in xs
]

```
